# Fabrica Biblia Reels

GPU gratuit (Colab T4 sau Kaggle GPU). Notebook-ul poate fi oprit și reluat: versetele gata nu se refac.

1. Runtime → Change runtime type → **T4** (în Kaggle: Settings → Accelerator → GPU).
2. Rulează celulele în ordine.
3. Descarcă folderul `biblia-reels/factory/output` și publică-l cum scrie în `biblia-reels/README.md`.

Pentru o probă, în celula de rulare pune `--limit 1` în loc de `--limit 0`.

```python
import os, subprocess
REPO = "https://github.com/gpeexportautomobile-cmyk/buton-sunet-ios.git"
BRANCH = "main"
FALLBACK = "cursor/biblia-reels-ff92"
if not os.path.exists("biblia-reels/factory/run_factory.py"):
    code = subprocess.call(["git", "clone", "--depth", "1", "--branch", BRANCH, REPO, "src"])
    if code != 0:
        subprocess.check_call(["git", "clone", "--depth", "1", "--branch", FALLBACK, REPO, "src"])
    os.chdir("src")
print(os.getcwd())
assert os.path.exists("biblia-reels/factory/run_factory.py")
```

```python
import os
root = "biblia-reels/factory/output"
for dirpath, dirnames, filenames in os.walk(root):
    dirnames[:] = [name for name in dirnames if name not in {"voices", "state"}]
    for name in sorted(filenames):
        if name.endswith((".webp", ".m4a", ".json")):
            print(os.path.join(dirpath, name))
```
